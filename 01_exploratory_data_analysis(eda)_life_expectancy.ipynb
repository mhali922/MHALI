# Machine Learning Lifecycle

A **machine learning project** follows a **structured lifecycle** that transforms **raw data** into **predictive models**.  
This process ensures that data is **properly understood, prepared, and used** to build **reliable machine learning systems**.

The **typical lifecycle of a machine learning project** is shown below.


# Machine Learning Lifecycle

| Step | Description |
|-----|-------------|
| **1. Data Collection** | Gather raw data from different sources such as databases, APIs, surveys, sensors, logs, or public datasets. The quality and relevance of collected data strongly influence the final model performance. |
| **2. EDA (Exploratory Data Analysis)** | Explore and understand the dataset by analyzing distributions, detecting missing values, identifying outliers, and examining relationships between variables using statistics and visualizations. |
| **3. Data Cleaning** | Improve data quality by handling missing values, removing duplicate records, correcting inconsistent entries, fixing data types, and filtering noisy or invalid data. |
| **4. Feature Engineering** | Transform raw variables into meaningful features that help models learn better patterns. This may include encoding categorical variables, scaling numerical features, creating new variables, or aggregating information. |
| **5. Model Training** | Apply machine learning algorithms to the prepared dataset so that the model learns patterns and relationships between input features and the target variable. |
| **6. Evaluation** | Assess the performance of the trained model using appropriate metrics such as accuracy, precision, recall, F1-score, RMSE, or R² depending on the problem type. |
| **7. Deployment** | Integrate the trained model into real-world systems such as web applications, APIs, cloud platforms, or mobile apps so it can make predictions on new incoming data. |
| **8. Monitoring** | Continuously track the model’s performance after deployment to detect issues such as data drift, concept drift, or declining accuracy. Models may need retraining when new data becomes available. |

### What are Quartiles?

Quartiles divide a dataset into **four equal parts**.

Life expectancy values:

55, 60, 62, 65, 68, 70, 72, 74, 76

Total values = **9**


### 50% (Median)

The **median** is the **middle value**.

Here, the **5th value** is the middle.

Median = **68**

Meaning:
- 4 values are **below 68**
- 4 values are **above 68**

Meaning: about **50% of the values are below ~68**.


### 25% (First Quartile / Q1)

Look at the **lower half** of the data:

55, 60, 62, 65

The middle of these values is around **60–62**.

So **Q1 ≈ 61**

Meaning: about **25% of the values are below ~61**.


### 75% (Third Quartile / Q3)

Look at the **upper half** of the data:

70, 72, 74, 76

The middle of these values is around **72–74**.

So **Q3 ≈ 73**

Meaning: about **75% of the values are below ~73**.

### Standard Deviation (std)

Standard deviation shows **how much the values vary from the average (mean)**.

### Formula

$\sigma = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \mu)^2}$

### Example (Life Expectancy)

Life expectancy values:

55, 60, 62, 65, 68, 70, 72, 74, 76

Number of observations → **n = 9**



### Step 1: Calculate the Mean

$$
\text{Mean} = \frac{55 + 60 + 62 + 65 + 68 + 70 + 72 + 74 + 76}{9}
$$


Mean ≈ **66.9**



### Step 2: Difference from Mean

| Value | Difference from Mean |
|------|----------------------|
| 55 | -11.9 |
| 60 | -6.9 |
| 62 | -4.9 |
| 65 | -1.9 |
| 68 | 1.1 |
| 70 | 3.1 |
| 72 | 5.1 |
| 74 | 7.1 |
| 76 | 9.1 |



### Step 3: Square the Differences

| Value | (x - mean)² |
|------|-------------|
| 55 | 141.6 |
| 60 | 47.6 |
| 62 | 24.0 |
| 65 | 3.6 |
| 68 | 1.2 |
| 70 | 9.6 |
| 72 | 26.0 |
| 74 | 50.4 |
| 76 | 82.8 |



### Step 4: Average of Squared Differences

Sum ≈ **386.8**


$$
\frac{386.8}{9} \approx 42.98
$$



### Step 5: Square Root

$$
\sqrt{42.98} \approx 6.55
$$

### Final Result

**Standard Deviation ≈ 6.55**

### Meaning

The life expectancy values vary about **±6.5 years around the average (66.9 years)**.

### Outliers

Outliers are **extreme values that are much higher or lower than most data points** in a dataset.

Outliers can affect statistical analysis and machine learning models.

We can detect outliers using **boxplots**.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(15,10))
for i, col in enumerate(numerical_cols.columns):
    plt.subplot(5,4,i+1)
    sns.boxplot(x=life_expectancy_df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

### Outlier Analysis

The boxplots show the presence of outliers in several numerical columns.

- **Year:** No significant outliers are observed.
- **Life expectancy:** A few lower values appear as outliers.
- **Adult Mortality:** Many high-value outliers are present.
- **infant deaths:** Several extreme outliers are visible.

These outliers represent unusually high or low values compared to the rest of the data.

### Univariate Analysis

**Univariate analysis** means analyzing **one variable at a time** to understand its characteristics.

- **Distribution** of the variable
- **Central tendency** (mean, median)
- **Spread** (min, max, standard deviation)
- **Outliers**

Examples of plots used:
- **Histogram** for numerical variables
- **Boxplot** for detecting outliers
- **Count plot / Bar chart** for categorical variables

### Bivariate Analysis

Bivariate analysis studies the relationship between **two variables**.

Examples:
- Scatter plot → relationship between GDP and Life expectancy
- Box plot → Life expectancy across Developed and Developing countries

#### Numerical vs Numerical (Scatter Plot)

```python
sns.scatterplot(
    x=life_expectancy_df["GDP"],
    y=life_expectancy_df["Life expectancy "]
)
plt.show()
```


The scatter plot shows the relationship between **GDP** and **Life expectancy**.

- Countries with **higher GDP generally tend to have higher life expectancy**.
- Many countries with **low GDP show a wide range of life expectancy values**.
- A few **high-GDP countries appear as outliers on the right side** of the plot.

This suggests a **positive relationship between GDP and life expectancy**.

#### Categorical vs Numerical (Box Plot)

```python
sns.boxplot(
    x=life_expectancy_df["Status"],
    y=life_expectancy_df["Life expectancy "]
)
plt.show()
```


The boxplot compares **Life expectancy** between **Developed** and **Developing** countries.

- **Developed countries** generally have **higher life expectancy**.
- The **median life expectancy** for developed countries is around **78–80 years**.
- **Developing countries** show a **lower median life expectancy** around **68–70 years**.
- Developing countries also show **greater variability** and some **low-value outliers**.

This suggests that a country's **development status is strongly associated with life expectancy**.

#### Numerical vs Numerical with a Categorical grouping (hue).

```python
sns.scatterplot(
    x="GDP",
    y="Life expectancy ",
    hue="Status",
    data=life_expectancy_df
)

plt.show()
```


The scatter plot shows the relationship between **GDP** and **Life expectancy**, with colors representing **country status**.

- **Developed countries (orange)** generally have **higher life expectancy** (around 75–85 years).
- **Developing countries (blue)** show a **wider range of life expectancy**, including many lower values.
- As **GDP increases**, life expectancy tends to **increase and stabilize**.
- Most **high-GDP countries are developed countries**.

This suggests that **economic development and GDP are positively associated with life expectancy**.
