### Applied AI Example: Complaint priority function

```python
def detect_priority(complaint_text):
    complaint_text = complaint_text.lower()
    if "urgent" in complaint_text or "blocked" in complaint_text:
        return "High"
    else:
        return "Normal"

print(detect_priority("My card is blocked"))
print(detect_priority("I want to know my balance"))
```

## 13. Combining Lists, Loops, Conditions, and Functions

Now let us process multiple complaints.

```python
complaints = [
    "My card is blocked and I need urgent help",
    "How can I update my mobile number?",
    "Loan payment failed",
    "I cannot login to mobile app"
]

def detect_priority(complaint_text):
    complaint_text = complaint_text.lower()
    if "urgent" in complaint_text or "blocked" in complaint_text or "cannot" in complaint_text:
        return "High"
    else:
        return "Normal"

for complaint in complaints:
    priority = detect_priority(complaint)
    print(complaint, "=>", priority)
```

## 16. Creating a Simple Dataset

Let us create a small customer complaint dataset.

```python
import pandas as pd

data = {
    "customer_id": [101, 102, 103, 104, 105, 106],
    "complaint_text": [
        "My card is blocked",
        "Mobile app is not working",
        "Loan payment issue",
        "ATM cash not received",
        "Credit card bill is wrong",
        "I need urgent help with my account"
    ],
    "product": ["Card", "Mobile App", "Loan", "ATM", "Card", "Account"],
    "channel": ["Call Center", "Mobile App", "Branch", "ATM", "Mobile App", "Call Center"],
    "status": ["Open", "Open", "Closed", "Open", "Closed", "Open"]
}

df = pd.DataFrame(data)
df
```

## 19. Creating a New Column

We can create a new column using Python logic.

Let us create a priority column.

```python
def detect_priority(complaint_text):
    complaint_text = complaint_text.lower()
    if "urgent" in complaint_text or "blocked" in complaint_text or "not working" in complaint_text:
        return "High"
    else:
        return "Normal"

df["priority"] = df["complaint_text"].apply(detect_priority)
df
```

## 20. Filtering Data

Filtering means selecting specific records.

```python
# Show only open complaints
open_complaints = df[df["status"] == "Open"]
open_complaints
```

## 24. From Data Analysis to Agentic AI Use Case

Agentic AI can take actions step by step.

Example workflow:
1. Read complaint
2. Detect issue type
3. Detect priority
4. Assign team
5. Draft response
6. Update tracker
7. Notify customer care team

Python helps connect these steps.

```python
def assign_team(product):
    if product == "Card":
        return "Card Operations"
    elif product == "Loan":
        return "Loan Operations"
    elif product == "Mobile App":
        return "Digital Banking"
    elif product == "ATM":
        return "ATM Support"
    else:
        return "Customer Care"

df["assigned_team"] = df["product"].apply(assign_team)
df
```
